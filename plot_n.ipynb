# Análisis de densidades de número en una simulación de neutrinos

Este notebook muestra cómo:

- Leer un archivo `HDF5` con resultados de una simulación
- Calcular promedios espaciales de densidades de número
- Graficar la evolución temporal de cada componente

Usaremos Python con las librerías `numpy`, `h5py` y `matplotlib`.


```python
# Librerías para análisis de datos y visualización
import os
import h5py
import numpy as np
import matplotlib.pyplot as plt
```

## Cargar archivo de simulación

Primero definimos la ruta donde se encuentra el archivo `allData.h5`, y lo abrimos en modo lectura.


```python
# Definir la ruta del directorio donde está el archivo
ruta_directorio = '/workspaces/Emu/Exec'

# Construir la ruta completa al archivo
ruta_archivo_h5 = os.path.join(ruta_directorio, 'allData.h5')

# Abrir el archivo y leer los datos dentro de un bloque 'with'
with h5py.File(ruta_archivo_h5, 'r') as archivo_h5:
    
    # Neutrinos
    dens_ee      = np.array(archivo_h5['/N00_Re(1|ccm)'][:])
    dens_uu      = np.array(archivo_h5['/N11_Re(1|ccm)'][:])
    dens_eu      = np.sqrt(np.array(archivo_h5['/N01_Re(1|ccm)'][:])**2 +
                           np.array(archivo_h5['/N01_Im(1|ccm)'][:])**2)
    
    # Antineutrinos
    dens_bar_ee  = np.array(archivo_h5['/N00_Rebar(1|ccm)'][:])
    dens_bar_uu  = np.array(archivo_h5['/N11_Rebar(1|ccm)'][:])
    dens_bar_eu  = np.sqrt(np.array(archivo_h5['/N01_Rebar(1|ccm)'][:])**2 +
                           np.array(archivo_h5['/N01_Imbar(1|ccm)'][:])**2)

    # Tiempo
    tiempo_s = np.array(archivo_h5['/t(s)'][:])
```

## Calcular promedio espacial

El archivo contiene un valor por celda para cada instante de tiempo. Vamos a promediar sobre todas las celdas espaciales para obtener una evolución temporal de las densidades promedio.


```python
# Promediar sobre todas las dimensiones excepto el tiempo (eje 0)
prom_ee       = np.mean(dens_ee,      axis=tuple(range(1, dens_ee.ndim)))
prom_uu       = np.mean(dens_uu,      axis=tuple(range(1, dens_uu.ndim)))
prom_eu       = np.mean(dens_eu,      axis=tuple(range(1, dens_eu.ndim)))
prom_bar_ee   = np.mean(dens_bar_ee,  axis=tuple(range(1, dens_bar_ee.ndim)))
prom_bar_uu   = np.mean(dens_bar_uu,  axis=tuple(range(1, dens_bar_uu.ndim)))
prom_bar_eu   = np.mean(dens_bar_eu,  axis=tuple(range(1, dens_bar_eu.ndim)))
```

## Visualizar los resultados

Ahora graficamos la evolución temporal de las densidades promedio para cada componente:

- **Neutrinos**:  $n_{ee}$, $n_{\mu\mu}$, $n_{e\mu}$
- **Antineutrinos**:  $\bar{n}_{ee}$, $\bar{n}_{\mu\mu}$, $\bar{n}_{e\mu}$

```python
# Crear la figura y el eje
figura, eje = plt.subplots(figsize=(10, 6))

# Graficar cada curva
eje.plot(tiempo_s, prom_ee,       label=r'$n_{ee}$')
eje.plot(tiempo_s, prom_uu,       label=r'$n_{\mu\mu}$')
eje.plot(tiempo_s, prom_eu,       label=r'$n_{e\mu}$')
eje.plot(tiempo_s, prom_bar_ee,   label=r'$\bar{n}_{ee}$')
eje.plot(tiempo_s, prom_bar_uu,   label=r'$\bar{n}_{\mu\mu}$')
eje.plot(tiempo_s, prom_bar_eu,   label=r'$\bar{n}_{e\mu}$')

# Etiquetas
eje.set_xlabel(r'Tiempo $t$ [s]')
eje.set_ylabel(r'Densidad $\langle n \rangle$ [$\mathrm{cm}^{-3}$]')
eje.legend(ncol=2, framealpha=0.0)
eje.grid(True)

# Mostrar la figura
plt.show()
```
